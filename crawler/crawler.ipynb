```python
import requests
from bs4 import BeautifulSoup

URL = "https://www.auckland.ac.nz/en/study/study-options/find-a-study-option/master-of-information-technology-minfotech.html"

headers = {
    "User-Agent": "UoA-Student-Navigator/1.0 (Educational Project)"
}

response = requests.get(URL, headers=headers)

print(f"Status Code: {response.status_code}")

html = response.text

print(html[:500])
```

```python
with open("../data/raw_html/master_it.html", "w", encoding="utf-8") as f:
    f.write(html)
```

```python
soup = BeautifulSoup(response.text, "lxml")

print(soup.title.text)
```

```python
main = soup.find("main", id="main")

# Remove scripts/styles
for tag in main(["script", "style", "noscript"]):
    tag.decompose()

```

```python
title = main.find("h1").get_text(strip=True)

headings = [
    h.get_text(strip=True)
    for h in main.find_all(["h2", "h3"])
]
```

```python
headings
```

```python
links = []

for a in main.find_all("a", href=True):
    href = a["href"]
    text = a.get_text(strip=True)

    links.append({
        "text": text,
        "href": href
    })
```

```python
links
```

```python
text = main.get_text(separator="\n", strip=True)
```

```python
text
```

```python
# saving the page

from pathlib import Path

Path("data").mkdir(exist_ok=True)

with open("../data/cleaned/master_it.txt", "w", encoding="utf-8") as f:
    f.write(text)
```

```python
import json

document = {
    "url": URL,
    "title": soup.title.text.strip(),
    "text": text,
    "headings":headings,
    "links":links
}

with open("../data/cleaned/master_it.json", "w", encoding="utf-8") as f:
    json.dump(document, f, indent=2, ensure_ascii=False)
```
